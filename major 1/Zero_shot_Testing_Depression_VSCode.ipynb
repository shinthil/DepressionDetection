```python
import numpy as np
from sentence_transformers import SentenceTransformer
from sklearn.model_selection import LeaveOneOut
from sklearn.metrics import classification_report
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping

# Step 1: Prepare data
pids = list(participant_texts_cleaned.keys())
texts = [participant_texts_cleaned[pid] for pid in pids]
y = np.array([labels[pid] for pid in pids])

# Step 2: Generate embeddings using MiniLM
encoder = SentenceTransformer('all-MiniLM-L6-v2')
X = encoder.encode(texts)

# Step 3: LOOCV Setup
loo = LeaveOneOut()
y_true, y_pred = [], []

# Step 4: Model builder function
def build_model():
    model = Sequential([
        Input(shape=(384,)),
        Dense(64, activation='relu'),
        Dropout(0.2),
        Dense(1, activation='sigmoid')
    ])
    model.compile(loss='binary_crossentropy', optimizer=Adam(0.001), metrics=['accuracy'])
    return model

# Step 5: LOOCV Loop
for train_idx, test_idx in loo.split(X):
    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]

    # Build & train model
    clf = build_model()
    clf.fit(X_train, y_train, epochs=50, batch_size=2, verbose=0,
            callbacks=[EarlyStopping(patience=5, restore_best_weights=True, monitor='loss')])

    # Predict
    pred_prob = clf.predict(X_test, verbose=0)[0][0]
    pred_label = int(pred_prob > 0.5)

    y_true.append(y_test[0])
    y_pred.append(pred_label)

# Step 6: Final Report
print("📊 Final LOOCV Classification Report:\n")
print(classification_report(y_true, y_pred, digits=2, zero_division=0))
```
