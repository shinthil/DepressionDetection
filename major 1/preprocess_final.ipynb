```python
# Cell 1: Imports and Setup
import os
import pandas as pd

input_dir = 'dataset'
output_dir = 'cleaned data'
os.makedirs(output_dir, exist_ok=True)

remove_phrases = [
    "hello", "hi", "hey", "goodbye", "bye",
    "how are you", "i'm good", "nice to meet you",
    "thanks", "thank you", "you too", "okay", "cool", "awesome"
]
filler_words = ["um", "uh", "ah", "hmm", "huh"]

def is_informative(text):
    text = str(text).lower().strip()
    if text in filler_words or len(text) <= 1:
        return False
    for phrase in remove_phrases:
        if phrase in text:
            return False
    return True
```

```python
# Cell 2: Preprocessing Function
def preprocess_transcript(file_path):
    try:
        # Explicitly specify tab separator for transcript files
        df = pd.read_csv(file_path, sep='\t', engine='python')
        df.columns = [col.lower().strip() for col in df.columns]

        if 'speaker' not in df.columns or 'value' not in df.columns:
            print(f"[!] Skipping: Missing 'speaker' or 'value' in {file_path}")
            print(f" Columns found: {df.columns.tolist()}")
            return None

        # Keep only relevant speakers and informative utterances
        df = df[df['speaker'].isin(['Ellie', 'Participant'])]
        df = df[df['value'].apply(is_informative)]

        # Only keep 'speaker' and 'value' columns
        return df[['speaker', 'value']]

    except Exception as e:
        print(f"[!] Error processing {file_path}: {e}")
        return None
```

```python
# Cell 3: Process All Files and Save Cleaned Data
processed_count = 0

for root, dirs, files in os.walk(input_dir):
    for fname in files:
        if "_transcript.csv" not in fname.lower():
            continue

        file_path = os.path.join(root, fname)
        module_name = os.path.splitext(fname)[0].split('_')[0]

        cleaned_df = preprocess_transcript(file_path)
        if cleaned_df is not None:
            output_file = os.path.join(output_dir, f"{module_name}_TRANSCRIPT_cleaned.csv")
            cleaned_df.to_csv(output_file, index=False)
            processed_count += 1
            print(f"[✓] Processed {fname} → {module_name}_TRANSCRIPT_cleaned.csv")

print(f"\n✅ Finished preprocessing {processed_count} files.")
```
